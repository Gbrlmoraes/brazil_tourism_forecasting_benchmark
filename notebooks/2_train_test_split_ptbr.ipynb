# Divisão Treino/Teste

```python
import os

import pandas as pd

ROOT_PATH = os.path.dirname(os.getcwd())
DATA_FOLDER_PATH = os.path.join(ROOT_PATH, 'data', 'processed')
```

```python
df = pd.read_parquet(os.path.join(DATA_FOLDER_PATH, 'cleaned_arrivals_ptbr.parquet'))

df.head()
```

## Divisão por Tempo

```python
print(
    f'Período completo: {df["date"].min().strftime("%d %B %Y")} a {df["date"].max().strftime("%d %B %Y")}'
)
```

```python
# Dados até dezembro de 2024 para treinamento
train_df = df[df['date'] < '2025-01-31']
print(
    f'Período do conjunto de treino: {train_df["date"].min().strftime("%d %B %Y")} a {train_df["date"].max().strftime("%d %B %Y")}'
)
train_df.tail()
```

```python
# Dados de 2025 em diante para teste
test_df = df[df['date'] >= '2025-01-31']
print(
    f'Período do conjunto de teste: {test_df["date"].min().strftime("%d %B %Y")} a {test_df["date"].max().strftime("%d %B %Y")}'
)
test_df.tail()
```

## Saída

```python
train_df.to_parquet(os.path.join(DATA_FOLDER_PATH, 'train_ptbr.parquet'), index=False)
test_df.to_parquet(os.path.join(DATA_FOLDER_PATH, 'test_ptbr.parquet'), index=False)
```
